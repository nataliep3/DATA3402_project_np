```python
# using a function to check for outliers using IQR
def detect_outliers_iqr(data):
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data < lower_bound) | (data > upper_bound)]
    return outliers
```

```python
# table showing a summary of each feature
feature_info = {}

for column in df.columns:
  feature_info[column] = {}

  # feature type sorting
  if df[column].dtype == 'int64' or df[column].dtype == 'float64':
    feature_info[column]['Type'] = 'Numerical'
    feature_info[column]['Range'] = f"{df[column].min()} - {df[column].max()}"

  # totaling missing values for each feature
  feature_info[column]['Missing Values'] = df[column].isnull().sum()

  # totaling outliers
  if feature_info[column]['Type'] == 'Numerical':
    outliers = detect_outliers_iqr(df[column])
    feature_info[column]['Outliers'] = len(outliers)

# actually outputting the table
feature_table = pd.DataFrame(feature_info).T
display(feature_table)
```

```python
# checking what the outliers are in the 'PCOS_Diagnosis' feature
df_diag = df['PCOS_Diagnosis']
detect_outliers_iqr(df_diag)
```
